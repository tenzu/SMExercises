```python
import os
import pandas as pd
import random


path = "./"
files = os.listdir(path)


# 定义读入单个文件生成 DataFrame 的函数
def readFile(fileName):
    df = pd.read_csv(fileName, sep="\t")
    return df


# 使用readFile函数读入 YNquestions.txt 文件
df = readFile("YNquestions.txt")


# 定义随机抽题函数
def randomSelect(df, n):
    random_indices = random.sample(range(len(df)), n)
    # 提取随机行
    random_rows = df.iloc[random_indices]
    # 将随机行数据保存到新的 DataFrame
    random_df = pd.DataFrame(random_rows)
    # 将random_df的index重置，序号从1开始
    random_df.reset_index(drop=True, inplace=True)
    random_df.index = random_df.index + 1
    # 输出每行的内容
    for index, row in random_df.iterrows():
        # index补齐为2位数
        index = str(index).zfill(2)
        # print(f"( {row['答案']} )\t{index}：{row['题目']}")
        print(f"(   )\t{index}：{row['题目']}")


# 手动输入n的值，n为小于200的正整数，n正确时调用randomSelect函数，输入q退出
while True:
    n = input("请输入抽题数量（小于200的正整数），输入q退出：")
    if n == "q":
        break
    try:
        n = int(n)
        if n <= 0 or n > 200:
            print("请输入小于200的正整数！")
        else:
            # 询问是否指定随机种子，输入数字时使用该数字作为随机种子，否则使用默认随机种子
            seed = input("请输入随机种子（数字），不输入则使用默认随机种子：")
            if seed.isdigit():
                random.seed(int(seed))
            else:
                random.seed()
            randomSelect(df, n)
    except ValueError:
        print("请输入一个有效的数字！")
```
